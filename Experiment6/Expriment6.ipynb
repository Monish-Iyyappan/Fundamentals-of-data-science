```python
import pandas as pd
import matplotlib.pyplot as plt

# Create website traffic data
data = {
    "Date": [
        "2026-02-01",
        "2026-02-02",
        "2026-02-03",
        "2026-02-04",
        "2026-02-05",
        "2026-02-06",
        "2026-02-07"
    ],
    "Visitors": [180, 220, 195, 260, 240, 310, 285]
}

# Create DataFrame
df = pd.DataFrame(data)

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

# Set Date as index
df.set_index("Date", inplace=True)

# Display the data
print("Website Traffic Data:")
print(df)

# Display basic information
print("\nStatistical Summary:")
print(df.describe())

# Calculate average visitors
print("\nAverage Website Visitors:", df["Visitors"].mean())

# Find maximum visitors
print("Maximum Website Visitors:", df["Visitors"].max())

# Find minimum visitors
print("Minimum Website Visitors:", df["Visitors"].min())

# Plot the time series
plt.figure(figsize=(10, 5))

plt.plot(
    df.index,
    df["Visitors"],
    marker="o"
)

plt.title("Time Series Analysis of Website Traffic")
plt.xlabel("Date")
plt.ylabel("Number of Visitors")
plt.xticks(rotation=45)

plt.grid(False)
plt.tight_layout()
plt.show()
```
