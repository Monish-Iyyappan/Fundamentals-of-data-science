```python
import pandas as pd
import matplotlib.pyplot as plt

# Create sample dataset
data = {
    "ProductID": [101, 102, 103, 104, 105],
    "Quantity": [10, 5, 8, 12, 7],
    "Price": [100.50, 200.75, 150.00, 50.25, 300.00]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display dataset
print("DATASET:")
print(df)

# Dataset information
print("\nDATASET INFORMATION:")
df.info()

# Summary statistics
print("\nSUMMARY STATISTICS:")
print(df.describe())

# Check for duplicates
print("\nDUPLICATES:")
print(df.duplicated().sum())

# Histogram without grid
df.hist(bins=10, figsize=(10, 6))
plt.grid(False)
plt.show()
```
