```python
#experiment 1.a

import pandas as pd
import matplotlib.pyplot as plt 
data = {
    'Year': list(range(2010, 2021)),
    'Job Postings': [95, 140, 225, 340, 510, 720, 980, 1320, 1810, 2450, 3250]
}
df = pd.DataFrame(data)
plt.plot(df['Year'], df['Job Postings'], marker='o') 
plt.title('Trend of Data Science Job Postings')
plt.xlabel('Year')
plt.ylabel('Number of Job Postings') 
plt.show()
```

```python
#experiment 1.b

import matplotlib.pyplot as plt
roles = ['Security Analyst', 'Penetration Tester', 'Security Architect', 'Incident Responder', 'Compliance Officer']
counts = [420, 180, 150, 210, 90]
plt.bar(roles, counts) 
plt.title('Distribution of Data Science Roles')
plt.xlabel('Role') 
plt.ylabel('Count') 
plt.show()
```
